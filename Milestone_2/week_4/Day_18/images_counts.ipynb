```python
base_path = r'/content/drive/MyDrive/AI_Vision_Extract_Nov25/data/COCO2017/train2017'
```

```python
import os
if 'base_path' in locals() or 'base_path' in globals():
    if os.path.exists(base_path):
        file_count = 0
        for item in os.listdir(base_path):
            item_path = os.path.join(base_path, item)
            if os.path.isfile(item_path):
                file_count += 1
        print(f"Number of files: {file_count}")
    else:
        print(f"Error: The path '{base_path}' does not exist.")
else:
    print("Error: 'base_path' variable is not defined. Please define it in a previous cell.")
```
